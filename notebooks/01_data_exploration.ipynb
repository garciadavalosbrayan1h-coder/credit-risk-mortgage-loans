¿Qué dimensiones tienen las dos muestras?
El archivo de originación de credito tiene una dimensión de: 31 columnas la de desempeño tiene una dimensión de 35 columnas

¿Por qué cinco faltantes no bastan para eliminar una columna del archivo completo?
Por que esa columna puede tener a proposito esas 5 columnas vacias.

¿Por qué todavía no conviene reemplazar los faltantes por cero?
Porque no sabemos la naturaleza de los datos y podríamos crear sesgo

Dimensiones: originación tiene 5 filas × 31 columnas y desempeño 5 filas × 35 columnas. La dimensión incluye filas y columnas.

Cinco faltantes: te refieres a cinco filas vacías dentro de una columna. Podrían estar vacías por una razón válida, pero el argumento principal es que solo revisamos cinco registros; otros registros podrían contener valores.

Reemplazar por cero: correcto. Cero es un valor concreto y puede tener un significado distinto de «dato desconocido» o «no aplica». Usarlo podría distorsionar el análisis.

```python
orig = pd.read_csv(
    orig_path,
    sep="|",
    header=None,
    dtype="string",
    usecols=[0, 9, 10, 11, 12, 19]
).rename(columns={
    0: "credit_score",
    9: "original_dti",
    10: "original_upb",
    11: "original_ltv",
    12: "original_interest_rate",
    19: "loan_id"
})

print("Filas y columnas:", orig.shape)
print("Identificadores faltantes:", orig["loan_id"].isna().sum())
print("Identificadores duplicados:", orig["loan_id"].duplicated().sum())
```

```python
for column in ["credit_score", "original_dti", "original_ltv"]:
    print(f"\nColumna: {column}")
    print(orig[column].value_counts(dropna=False).head(10))
```

```python
print(
    "Puntajes con código 9999:",
    orig["credit_score"].eq("9999").sum()
)

print(
    "Puntajes vacíos:",
    orig["credit_score"].isna().sum()
)
```

```python
orig_clean = orig.copy()

orig_clean["credit_score"] = pd.to_numeric(
    orig_clean["credit_score"].replace("9999", pd.NA),
    errors="raise"
).astype("Int64")

print("Puntajes faltantes:", orig_clean["credit_score"].isna().sum())
print("Tipo:", orig_clean["credit_score"].dtype)
print("Filas conservadas:", len(orig_clean))
```

```python
print("Puntaje mínimo:", orig_clean["credit_score"].min())
print("Puntaje máximo:", orig_clean["credit_score"].max())

outside_range = (
    (orig_clean["credit_score"] < 300)
    | (orig_clean["credit_score"] > 850)
)

print("Puntajes fuera de 300–850:", outside_range.sum())
```

```python
score_quality = pd.Series({
    "total_creditos": len(orig_clean),
    "puntajes_disponibles": orig_clean["credit_score"].count(),
    "puntajes_faltantes": orig_clean["credit_score"].isna().sum(),
    "puntaje_minimo": orig_clean["credit_score"].min(),
    "puntaje_maximo": orig_clean["credit_score"].max()
})

display(score_quality)
```

```python
print(
    "Puntaje promedio:",
    round(orig_clean["credit_score"].mean(), 2)
)
```

```python
orig_clean["original_dti"] = pd.to_numeric(
    orig_clean["original_dti"],
    errors="raise"
).astype("Int64")

print("DTI disponibles:", orig_clean["original_dti"].count())
print("DTI faltantes:", orig_clean["original_dti"].isna().sum())
print("DTI mínimo:", orig_clean["original_dti"].min())
print("DTI máximo:", orig_clean["original_dti"].max())
```

```python
high_dti = orig_clean["original_dti"] > 40

print("Créditos con DTI > 40 %:", high_dti.sum())
print("Porcentaje del total:", round(high_dti.mean() * 100, 2))
```

¿cómo comprobarías si los créditos con DTI superior al 40 % presentaron más morosidad que los demás?

Para comparar los grupos:
1. Uniríamos originación y desempeño mediante loan_id.
2. Separaríamos los créditos en DTI > 40 % y DTI ≤ 40 %.
3. Definiríamos una misma medida de morosidad y un periodo de observación comparable.
4. Compararíamos el porcentaje de créditos con morosidad de cada grupo, no solo su cantidad, porque los grupos tienen tamaños distintos.

```python
print(
    orig["original_ltv"]
    .value_counts(dropna=False)
    .sort_index()
    .to_string()
)
```

```python
orig_clean["original_ltv"] = pd.to_numeric(
    orig_clean["original_ltv"],
    errors="raise"
).astype("Int64")

print("LTV disponibles:", orig_clean["original_ltv"].count())
print("LTV faltantes:", orig_clean["original_ltv"].isna().sum())
print("LTV promedio (%):", round(orig_clean["original_ltv"].mean(), 2))
print("LTV mediano (%):", orig_clean["original_ltv"].median())
```

```python
high_ltv = orig_clean["original_ltv"] > 80

print("Créditos con LTV > 80 %:", high_ltv.sum())
print("Porcentaje del total:", round(high_ltv.mean() * 100, 2))
```

```python
high_dti_ltv = (
    (orig_clean["original_dti"] > 40)
    & (orig_clean["original_ltv"] > 80)
)

print("Créditos que cumplen ambas condiciones:", high_dti_ltv.sum())
print("Porcentaje del total:", round(high_dti_ltv.mean() * 100, 2))
```

¿Por qué cumplir ambas condiciones podría ser relevante para analizar el riesgo, sin que eso signifique que el crédito necesariamente caerá en morosidad?´

R: Considero que es muy relevante ya que también podemos medir el riesgo que represetan clientes con alto porcentaje de DTI y LTV, me refiero que al menos en esta muestra representa casi el 7% de los 50000 clientes. Si el nivel, hipoteticamente fuera del 50% entonces esa empresa estaria tomando un gran riesgo o no necesariamente. Solo que debe tener en cuenta que tiene muchos clientes con casi el 50% de su ingreso a deudas y el prestamo vale el 80% de la hipoteca. Entonces no es ncesarimanete malo, y creo que se deben evaluar más variables para saber exactamente cual es la combinación de factores que nos indiquen la efectividad y término del prestamos, terminará siendo moroso o simplemente no pagara

DTI > 40 % significa que más del 40 % del ingreso mensual considerado se destina a deudas; no necesariamente casi el 50 %.
LTV > 80 % significa que el préstamo inicial supera el 80 % del valor de la propiedad.
Contamos créditos, no necesariamente clientes distintos.

Si hipotéticamente el 50 % de la cartera cumpliera ambas condiciones, habría mayor concentración en ese perfil. Eso justificaría examinarla con más atención, pero no bastaría para concluir que la empresa está tomando un riesgo excesivo. Necesitaríamos conocer el desempeño, otras características y las protecciones frente a pérdidas.

```python
orig_clean["original_upb"] = pd.to_numeric(
    orig_clean["original_upb"],
    errors="raise"
).astype("Int64")

print("Capitales disponibles:", orig_clean["original_upb"].count())
print("Capitales faltantes:", orig_clean["original_upb"].isna().sum())
print("Capital mínimo:", orig_clean["original_upb"].min())
print("Capital máximo:", orig_clean["original_upb"].max())
print("Capitales cero o negativos:", (orig_clean["original_upb"] <= 0).sum())
```

```python
total_upb = orig_clean["original_upb"].sum()

group_upb = orig_clean.loc[
    high_dti_ltv, "original_upb"
].sum()

print(f"Capital inicial total: ${total_upb:,.0f}")
print(f"Capital inicial del grupo: ${group_upb:,.0f}")
print(f"Participación en el capital: {group_upb / total_upb * 100:.2f}%")
```

```python
average_upb = orig_clean["original_upb"].mean()
group_average_upb = orig_clean.loc[
    high_dti_ltv, "original_upb"
].mean()

print(f"Capital inicial promedio general: ${average_upb:,.2f}")
print(f"Capital inicial promedio del grupo: ${group_average_upb:,.2f}")
```
